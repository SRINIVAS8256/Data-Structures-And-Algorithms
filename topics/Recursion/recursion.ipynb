```python
def f1(count):
    print(count)
    if count == 5:
        return count
    f1(count+1)
def main():
    count=0
    f1(count)
main()
```

```python
# print n times
def f1(i,n):
    if i>n:
        return
    print('hi',i)
    # i=i+1  or 
    f1(i+1,n)
f1(1,5)
```

```python
# 1 to N 
start=0
def f(n):
    global start
    if n==0:
        return
    print(start)
    start+=1
    if start == n:
        return
    f(n)
f(5)
    
    
```
