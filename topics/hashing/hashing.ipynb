```python
def h(check):
    arr=[1,2,3,4,5,6,1,2,34,4,5,76,1,2,3,4,5,6,1,0,0,0,1,1,1,2,2,2,3,4]
    size=max(arr)
    new_h=[0] * (size+1)
    for i in arr:
        new_h[i] =new_h[i]+1
    if check > max(arr):
        return -1
    return new_h[check]
```

```python
print(h(11))
```

```python
def char_hashing(check):
    arr=['a','b','a','b','c']
    max_val=0
    # finding max ascii value
    for i in arr:
        if ord(i) > max_val:
            max_val=ord(i)
    # creating new array of size max - 97
    char_arr=[0]*(max_val-97 + 1)

    #inserting values in new array
    for i in arr:
        char_arr[ord(i)-97]=char_arr[ord(i)-97] + 1
    if ord(check) > max_val:
        return -1
    return char_arr[ord(check)-97]
```

```python
char_hashing('d')
```
